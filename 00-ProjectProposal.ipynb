## Research Question

How do college students’ academic performance & GPA differ between those who has a high Artificial Intelligence(AI) usage category than those who do not or loosely use AI?

## Background and Prior Work

AI has become a major part of student's lives, with services such as LLMs becoming more popular during recent years. Our team members have had experience with AI being used in class projects, and have personally known people that heavily rely on AI to complete their school assignments. As mentioned in our current research question, we are curious about whether there is a significant difference in college student's GPA between college students who heavily rely on AI tools vs. college students who do not or rarely use AI tools for their school assignments.

Previous research work has investigated how the use of AI in higher education has reshaped how students interact with their academics. One review paper, published in the Frontiers of Psychology journal in 2025, mentioned that while AI tools offer a personalized learning experience, it also raises concerns over increased loneliness and stress.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) The review paper mentions how certain scholars believe that AI is a useful tool to increase academic performance by giving students a personalized learning experience, while other scholars believe that an overreliance on AI increases loneliness and reduces emotional intelligence which can negatively impact how students work together in group projects or in teamwork-based settings. Personalized learning experience, loneliness, and stress all can impact students' GPA and overall academic performance, which raises questions about if a significant difference exists in students' academic performance between those that heavily rely on AI and do not rely on AI.

Building on this, other research has examined how artificial intelligence can directly and indirectly influence academic performance in college students. One study, “Optimization of academic performance and mental health in college students through an AI-driven personalized intervention system,” found that students who used an AI-based system experienced a 10.28% increase in GPA, along with reduced stress and improved overall well-being.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) The study also found that students showed improvements in daily habits such as physical activity, sleep, and consistency in routines, which are all factors linked to better academic performance. Unlike tools such as ChatGPT that are typically used for completing assignments, this study applied AI to improve students’ habits, including exercise and mindfulness. This suggests that AI may influence GPA not only through academic assistance, but also through improvements in stress management, engagement, and study behaviors. However, the study also highlights that the effect of AI depends on how it is used. While some students may benefit from AI as a supportive learning tool, others who rely on it too heavily for completing assignments may not develop the same level of understanding or critical thinking. This is important for our project because it directly relates to our comparison between high and low AI usage, suggesting that differences in how frequently and how students use AI may result in measurable differences in college students’ GPA.
