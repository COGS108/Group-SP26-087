## Research Question

Is there a significant difference in GPA between college students with high AI tool usage and those with low or no AI tool usage? AI usage will be measured using hours of AI usage and scores of AI dependency.

## Background and Prior Work

AI has become a major part of students' lives, with services such as LLMs becoming more popular during recent years. Our team members have had experience with AI being used in class projects, and have personally known people that heavily rely on AI to complete their school assignments. As mentioned in our current research question, we are curious about whether there is a significant difference in college students' GPA between college students who heavily rely on AI tools vs. college students who do not or rarely use AI tools for their school assignments.

Previous research work has investigated how the use of AI in higher education has reshaped how students interact with their academics. One review paper, published in the Frontiers of Psychology journal in 2025, mentioned that while AI tools offer a personalized learning experience, it also raises concerns over increased loneliness and stress.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) The review paper mentions how certain scholars believe that AI is a useful tool to increase academic performance by giving students a personalized learning experience, while other scholars believe that an overreliance on AI increases loneliness and reduces emotional intelligence which can negatively impact how students work together in group projects or in teamwork-based settings. Personalized learning experience, loneliness, and stress all can impact students' GPA and overall academic performance, which raises questions about if a significant difference exists in students' academic performance between those that heavily rely on AI and do not rely on AI. The review also emphasized the rapid integration of AI into higher education, describing AI-driven technologies as increasingly “indispensable tools” for students in both academic and personal settings. Students commonly use AI-powered systems for assignments, research assistance, communication, and personalized learning experiences, showing how deeply AI has become embedded in modern college life. However, because the paper was mainly a review of existing research rather than a direct experimental study, it did not specifically compare GPA differences between students with different levels of AI usage. In addition, the study focused more broadly on AI use and student well-being instead of specifically looking at how students use generative AI tools like ChatGPT in their everyday coursework and assignments.

Building on this, other research has examined how artificial intelligence can directly and indirectly influence academic performance in college students. Another study, “Optimization of academic performance and mental health in college students through an AI-driven personalized intervention system,” found that students who used an AI-based system experienced a 10.28% increase in GPA, along with reduced stress and improved overall well-being.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) The study also found that students showed improvements in daily habits such as physical activity, sleep, and consistency in routines, which are all factors linked to better academic performance. Unlike tools such as ChatGPT that are typically used for completing assignments, this study applied AI to improve students’ habits, including exercise and mindfulness. This suggests that AI may influence GPA not only through academic assistance, but also through improvements in stress management, engagement, and study behaviors. However, the study mainly focused on a structured AI wellness intervention rather than the everyday academic use of generative AI tools such as ChatGPT for assignments or coursework. In addition, the study was conducted only among university students in China, which may limit how well the findings apply to students in other educational or cultural settings. The researchers also acknowledged that its design could not fully establish direct cause-and-effect relationships between AI usage and academic performance. Despite these limitations, the study highlights that the effects of AI may depend on how it is used. While some students may benefit from AI as a supportive learning tool, others who rely on it too heavily for completing assignments may not develop the same level of understanding or critical thinking. This directly relates to our project because we aim to examine whether differing levels of AI reliance are associated with measurable GPA differences among college students.

However, other research suggests that excessive reliance on AI tools may have negative effects on students’ learning and cognitive development. A study on the role of AI in education, "The cognitive paradox of AI in education: between enhancement and erosion" highlights the concept of “cognitive offloading”, where students depend on AI systems to complete tasks instead of actively engaging in problem-solving and critical thinking.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) This overreliance can reduce memory retention, lower cognitive engagement, and limit the development of higher-order thinking skills, even if AI improves efficiency in the short term. Additionally, the study suggests that while AI can enhance accessibility and support learning, it may also decrease motivation and independent reasoning when used too frequently. However, the study mainly focuses on the theoretical and cognitive impacts of AI in education rather than directly examining academic outcomes such as GPA differences between students with high and low AI usage. Moreover, much of the paper is based on existing literature and theoretical frameworks rather than long-term experimental studies in realistic classroom settings. These findings are important for our project because they suggest that students who rely too heavily on AI tools may have different academic outcomes than students who use AI more moderately or independently, which connects directly to our comparison of AI usage and GPA.

Overall, previous research shows both positive and negative effects of AI in education. While some studies suggest AI can improve learning and reduce stress, others argue that relying too heavily on AI may hurt critical thinking and cognitive engagement. Many existing studies also focus more on structured AI programs or theory rather than how students actually use tools like ChatGPT in everyday coursework. Our project aims to explore this further by comparing GPA differences between students with different levels of AI usage.

### Dataset #1 - Academic Outcomes & AI Dependency

Our dataset looks at relations between AI usage and dependency with academic grades which is useful for our project because it will help us explore whether or not there is association between the two. The Academic Outcomes & AI Dependency dataset contains 8000 observations and 26 variables. The important metrics in this dataset consist of `study_hours_per_day`, `uses_ai`, `ai_usage_time_minutes`, `ai_dependency_score`, `last_exam_score`, and `final_score`. The first metric, `study_hours_per_day` measures the amount of time students dedicate to studying each day, which is recorded in hours. A lower value indicates that the students do not study as much on average compared to those with a higher value in this category. The next variable is `uses_ai`, which is recorded in 0 and 1. 0 indicates that the student has not used AI for their studies while 1 indicates that they have. The next variable is `ai_usage_time_minutes` which is recorded in minutes. Similar to `study_hours_per_day`, this metric measures the amount of time students dedicate but for their AI chatbots instead of their studies. A higher value here indicates that users utilize AI more for their studies while a lower value indicates a lower usage of AI. The next variable, `ai_dependency_score`, is a self-report on how much the students think they depend on AI on a scale from 1-10, with 1 being a very low dependency and 10 being the highest. A higher value on the scale tells us that this particular student relies more heavily on AI tools that students who report a lower value on our scale. The last two variables, `last_exam_scores` and `final_score`, are similar as they both tell the scores the students have recieved on their last exam and also their final grade. The scores are measured from 0-100, which a score of 100 indicating better academic achievements. 

We do have some concern with the dataset, in particular that the dataset is not particularly clean, as it is missing values in a few columns and rows. Missing values may cause our data to slightly affect analysis, as there could be a chance that we are unable to find correlation if enough people in one category refuse to respond with meaningful data. Another concern we have is that the data is mostly self-reported. Implicit bias/poor memory may cause innacuracies within the data, as students may overestimate or underestimate in some categories like `study_hours_per_day`, `ai_usage_time_minutes`, or any other time recording variables. Along with that, ai_dependency is biased from the start as everyone has different ideas on what each value on the scale represents. For example, some people may rate themselves a 5 as they are middle of the pack within their friend group in ai usage, but this may be a 3 or a 7 to other people.

### Dataset #3 - AI Usage, Study Habits, and Academic Performance Among Students

This dataset looks at how generative AI usage may affect student academic performance and study habits. It contains around 50,000 observations and 16 variables related to GPA, studying behavior, AI usage, stress levels, and knowledge retention. This dataset is useful for our project because it directly includes measurements of both AI usage and academic outcomes, which can help us explore whether students who rely more heavily on AI perform differently in school compared to students who use AI less often.

There are several variables that are important for analyzing the relationship between AI usage and academic performance. `Pre_Semester_GPA` and `Post_Semester_GPA` show a student’s GPA before and after the semester on a 1.0–4.0 scale, which can help measure changes in academic performance over time, where values closer to 4.0 represent stronger academic performance. `Weekly_GenAI_Hours` tracks how many hours students spend using generative AI tools each week, while `Traditional_Study_Hours` records time spent studying without AI support. Another useful variable is `Perceived_AI_Dependency`, which reflects how strongly students believe they rely on AI tools for schoolwork or studying, where higher values indicate greater reliance on AI assistance. Looking at these variables together may help show whether increased AI usage changes study habits or academic outcomes.

The dataset also includes variables related to stress and learning retention. `Anxiety_Level_During_Exams` measures exam anxiety on a scale from 1 to 10, with higher scores representing higher stress levels during exams. `Skill_Retention_Score` measures how well students retained information after the semester ended, which may help show whether relying heavily on AI affects long-term understanding of material. These variables are useful because they provide additional insight beyond GPA alone and may help show whether heavy AI usage affects long-term understanding or student well-being.

One major limitation of this dataset is that many of the variables are self-reported, meaning students may not accurately report their study hours, AI usage, dependency, or anxiety levels. The dataset may also contain self-selection bias since students interested in AI or technology could be more likely to participate in the survey. In addition, while the dataset may show relationships between AI usage and academic performance, it cannot fully prove causation because other factors such as motivation, course difficulty, or previous academic ability could also affect the results.
