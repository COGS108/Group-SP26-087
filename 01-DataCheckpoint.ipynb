## Research Question

Is there a significant difference in GPA between students with high AI usage and students with low or no AI usage? AI usage will be measured using hours of AI usage and score of AI dependency.

## Background and Prior Work

AI has become a major part of student's lives, with services such as LLMs becoming more popular during recent years. Our team members have had experience with AI being used in class projects, and have personally known people that heavily rely on AI to complete their school assignments. As mentioned in our current research question, we are curious about whether there is a significant difference in college student's GPA between college students who heavily rely on AI tools vs. college students who do not or rarely use AI tools for their school assignments.

Previous research work has investigated how the use of AI in higher education has reshaped how students interact with their academics. One review paper, published in the Frontiers of Psychology journal in 2025, mentioned that while AI tools offer a personalized learning experience, it also raises concerns over increased loneliness and stress.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) The review paper mentions how certain scholars believe that AI is a useful tool to increase academic performance by giving students a personalized learning experience, while other scholars believe that an overreliance on AI increases loneliness and reduces emotional intelligence which can negatively impact how students work together in group projects or in teamwork-based settings. Personalized learning experience, loneliness, and stress all can impact students' GPA and overall academic performance, which raises questions about if a significant difference exists in students' academic performance between those that heavily rely on AI and do not rely on AI. The review also emphasized the rapid integration of AI into higher education, describing AI-driven technologies as increasingly “indispensable tools” for students in both academic and personal settings. Students commonly use AI-powered systems for assignments, research assistance, communication, and personalized learning experiences, showing how deeply AI has become embedded in modern college life. However, because the paper was mainly a review of existing research rather than a direct experimental study, it did not specifically compare GPA differences between students with different levels of AI usage. In addition, the study focused more broadly on AI use and student well-being instead of specifically looking at how students use generative AI tools like ChatGPT in their everyday coursework and assignments.

Building on this, other research has examined how artificial intelligence can directly and indirectly influence academic performance in college students. Another study, “Optimization of academic performance and mental health in college students through an AI-driven personalized intervention system,” found that students who used an AI-based system experienced a 10.28% increase in GPA, along with reduced stress and improved overall well-being.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) The study also found that students showed improvements in daily habits such as physical activity, sleep, and consistency in routines, which are all factors linked to better academic performance. Unlike tools such as ChatGPT that are typically used for completing assignments, this study applied AI to improve students’ habits, including exercise and mindfulness. This suggests that AI may influence GPA not only through academic assistance, but also through improvements in stress management, engagement, and study behaviors. However, the study mainly focused on a structured AI wellness intervention rather than the everyday academic use of generative AI tools such as ChatGPT for assignments or coursework. In addition, the study was conducted only among university students in China, which may limit how well the findings apply to students in other educational or cultural settings. The researchers also acknowledged that its design could not fully establish direct cause-and-effect relationships between AI usage and academic performance. Despite these limitations, the study highlights that the effects of AI may depend on how it is used. While some students may benefit from AI as a supportive learning tool, others who rely on it too heavily for completing assignments may not develop the same level of understanding or critical thinking. This directly relates to our project because we aim to compare GPA differences between students with high and low levels of AI usage in realistic academic settings.

However, other research suggests that excessive reliance on AI tools may have negative effects on students’ learning and cognitive development. A study on the role of AI in education, "The cognitive paradox of AI in education: between enhancement and erosion" highlights the concept of “cognitive offloading”, where students depend on AI systems to complete tasks instead of actively engaging in problem-solving and critical thinking.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) This overreliance can reduce memory retention, lower cognitive engagement, and limit the development of higher-order thinking skills, even if AI improves efficiency in the short term. Additionally, the study suggests that while AI can enhance accessibility and support learning, it may also decrease motivation and independent reasoning when used too frequently. However, the study mainly focuses on the theoretical and cognitive impacts of AI in education rather than directly examining academic outcomes such as GPA differences between students with high and low AI usage. Moreover, much of the paper is based on existing literature and theoretical frameworks rather than long-term experimental studies in realistic classroom settings. These findings are important for our project because they suggest that students who heavily rely on AI tools may not experience the same academic benefits as students who use AI more moderately or independently, which directly relates to our comparison of AI usage levels and GPA outcomes.

Overall, previous research shows both positive and negative effects of AI in education. While some studies suggest AI can improve learning and reduce stress, others argue that relying too heavily on AI may hurt critical thinking and cognitive engagement. Many existing studies also focus more on structured AI programs or theory rather than how students actually use tools like ChatGPT in everyday coursework. Our project aims to explore this further by comparing GPA differences between students with different levels of AI usage.
